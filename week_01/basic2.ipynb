## 기본문법

### 컨테이너 튜플
- 리스트와 비슷하지만, Immutable 자료형으로 새로운 요소를 추가, 삭제, 갱신할 수 없다.
- 항상 고정되거나 변경되어서는 안되는 데이터등을 저장하는데 사용할 수 있다.

```python
t = (0,1,2,3)
print(t, type(t))
t = 0,1,2,3
print(t, type(t))
a,b,c,d = t
print(a,b,c,d)
t = ((1,2), (3,4), (5,6)) # nested tuples
print(t, type(t))
a,b,c=t
print(a,b,c)
```

```python
i, j = 0,1
print(t[i], t[i][j])
print(t[i:j])
print(len(t))
print(t[1]+t[2]) # concatenate
print(t*3) # repeat
print(3 in t) # membership
t[0][0]=7 # error, as Tuple is immutable
```

### 컨테이너 딕션어리

- 사전과 같이 Key, Value 쌍을 요소로 갖는다.
- Key: immutable type

```python
d = {} # empty dictionary
print(d, type(d)) 
d = dict(a=1, b=2, c=3)
print(d, type(d)) 

print(d['a']) # indexing by key
d['a'] = -1 # update
print(d['a'])

print('------------')
print('a' in d) # membership test
print(d.keys())
print(d.values())
print(list(d.keys())) # keys list
print(list(d.values())) # values list
print(list(d.items())) # list of tuple (key, value)
```

```python
d = dict(a=1, b=2, c=3)
print(len(d)) # length
d['d'] = 4 # adding/changing
print(d)
del d['a']
print(d)
```

### 집합(Set) 자료형

- Set은 중복이 없는 컨테이너 자료형으로 순서도 없다.

```python
x=set({1,2,3,4})
print(x)
y=set(['cat', 'dog', 'tiger'])
print(y)
y = {'cat', 'dog', 'tiger'}
print(y)
print('-------------')
z = set() # empty set
z.add('fish')
print(z)
z.remove('fish')
print(z)
z.update({'lion', 'butterfly', 'bear'})
print(z)
z.clear()
print(z)
print('-------------')
x = {1,2,3,4,5}
y = {3,4,5,6,7}
z1 = x|y # union
print(z1)
z2 = x&y # intersection
print(z2)
z3 = x-y # difference
print(z3)
```

## 반복문, 조건문

### 반복문(if, while)

- 반복되는 루프를 만들기 위해 for 또는 while 문을 사용한다.
- 주로 컨테이너들 안에 들어있는 요소들을 하나씩 가져오는 방식으로 반복문이 사용되는 경우가 많다.

```python
for i in [0,1,2,3,4]: # tab or spaces (indentation)
    print('i=', i)
print('--------------')

for i in range(5):
    print(f'i= {i:2.2f}') # range, formatting
print('--------------')

for k, i in enumerate(range(5)):
    print(k, f'i= {i:2.2f}') # enumerate
print('--------------')

b = ['a', 1, 'b', 'dog', [0,1,2,3], {'a':1, 'b':5}]
for k, i in enumerate(b): 
    print(k, i) # container
print('--------------')

c = 'abcdefg'
for k, i in enumerate(c):
    print(k, i)
print('--------------')
import numpy as np
a = np.arange(3,15,2)
print(a)
for i in a:
    print('i=', i)
print('--------------')
```

```python
i = 1
total = 0
while i <= 20: # while loop
    total += i
    print(f'{i:4} total={total:6}')
    i += 1
```

### Bool type과 조건문(if)

```python
# bool type
print(1 == 2)
print(1 != 2)
print(not 1 == 2)
print(1>2)
print(1<2 and 2<4)
print(True and True)
print(True and False)
print(True or False)
print('------------')
flag = 2>4
print(flag)
if flag:  # if, elif, else
    print('True')
else:
    print('False')
print('------------')
x=2
if x < 0:
    print('x is less than 0')
elif x >= 0 and x < 5:
    print('x is in [0, 5)')
else:
    print('x >= 5')
```

### break/continue statements

- break: 루프를 빠져나오는 방법
- continue: 루프를 돌아가는 중에 해당 사이클에서 아래 코드를 실행하지 않고 다음루프로 가게한다. 주로 조건문과 함께 쓰인다.

```python
for i in np.arange(1,20,1):
    if i % 3 == 0:
        print(i, 'omitting')
    else:
        print(i, i % 3)
```

```python
for k, i in enumerate(range(100)):
    if i > 10:
        break
    print(k, i*10)
```
