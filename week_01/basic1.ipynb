## 기본문법

### 기본 데이터타입, 변수와 문자열

```python
x=2 # 정수형 데이터
y=3.14 # 실수형 데이터
z=1+3.4j # 복소수 데이터
a = True # Bool 데이터
b = False

s = 'Hello world!' # 문자열 데이터
t = None # Null
```

### 파이썬 기본 연산

```python
# 산술연산: +, -, *, /, **, %, // (기본 수치 연산자들)
x=1
y=3
z=x+y
print(x, y, z)

# 비교연산: ==, !=, <, >, <=, >= (기본 비교연산자들)
if z == 4:
    print('z is 4')
elif z != 4:
    print('z is not 4')

# 할당연산: =, +=, -=, *=, /=, %=, //=
x = x*3
x *= 3 # 위와 같은 의미
print(x)
```

```python
# 논리연산: and, or, not
a = True
b = False

if a or b:
    print('True')
else:
    print('False')
```

```python
# 멤버쉽: in, not in
x = [1, 2, 3]
a = 3 in x # True
print(a)

if 1 in x:
    print('Yes')
else:
    print('No')
```

## 문자열

- 파이썬에는 문자열을 표현하기 위해 단일인용부호 ' 또는 이중인용부호 " 를 사용한다
- 여러 라인을 사용하는 문자열은 ''' 이와같이 단일 인용부호 3개를 사용한다 (이중인용부호 3개를 사용해도 된다.)

```python
s = 'hello world' # single quotation
print(s)

t = "hello world" # double quotation
print(t)
print(t[0:3]) # sub-scripting
```

```python
s = '''
Hello world!
Is this the real world?
My heart leaps up when I behold a rainbow in the sky
'''
print(s)
```

### 문자열 포맷팅

- 문자열을 일정한 형식으로 만드는 것을 문자열 포맷팅이라고 한다. 파이썬에서는 3가지 문자열 포맷 방식을 제공한다.
- % formatting: c-style formatting
- string formatting: {}에 대응하는 것을 format()의 인자로 전달
- f-string: {}에 직접 변수를 입력 (가독성이 좋다)

```python
# % formatting
Loss = 0.001 ; y = 1000.
s='Loss= %10.4f, Value= %10.2f' % (Loss, y)
print(s)
```

```python
# String formatting
print('Loss= {:10.4f}, Value= {:10.2f}'.format(Loss, y))
```

```python
# f-string
print(f'Loss= {Loss:10.4f}, Value= {y:10.2f}')
```

## 컨테이너

### 파이썬 컨테이너 (Container)

- 데이터들을 담아놓을 수 있는 컨테이너: 리스트, 튜플, 딕셔너리 그리고 집합

#### 리스트(List) 자료형: 대표적 컨테이너

```python
a = [1,2,3,4,5]
print(a)
print(len(a)) # length
print(a[0]) # indexing
print(a[-1]) # reverse indexing
print('----------------------')

a[1] = 100
print(a)
print('----------------------')

a = [0,1]*3
print(a)
print(a[1:4]) # slicing
print(a[:3])
print(a[:-2])
print('----------------------')
a = [1,2,3,4,5,6,7,8]
print(a)
print(a[::2]) # step
print(a[2:])
print(a[2::2])
print('----------------------')
print(a[2:7])
print(a[2:7:2])
print(a[::-1]) # reverse using step -1
print(a[::-2]) # reverse using step -2
```

#### 리스트 자료형의 기본 메소드들

```python
a = [1,2,3,4,5,6]
print(dir(a)) # attributes of the object
print('----------------------')
print(a)
a.append(5) # add 5 into the list a
a.remove(1) # remove 1 from the list a
print(a)
a.insert(1,2) # insert
print(a)
print(a.index(6), a.count(2)) # index, count
print(max(a), min(a)) # max, min value
```

```python
a = [3,7,3,1,2,6]
print(a)
a.sort() # sorting
print(a)

a = [3,7,3,1,2,6]
sorted(a) # sorting (external function)
print(a)
```

```python
a = [3,7,3,1,2,6]
print(a)
print(a.pop(1), a) # pop: return and remove item i from the list
del a[0] # delete first one
del a[-1] # delete last one
print(a)
a[0]=1 # update first one
a[0:2]=[4,5] # update with slice
print(a)
```

#### 레퍼런싱, 리스트 복사

```python
a = [1,2,3]
b=a # referencing, alias
print(a is b)
print(a == b)
print(a, b)
a[0] = 100
print(a is b)
print(a == b)
print(a, b)
print(id(a), id(b)) # memory address
```

```python
# Shallow copy
a=[1,2,3]
b = a[:]
c = a.copy()
print(id(a), id(b), id(c))
print(a is b)
print(a==b)
print(a is c)
print(a==c)
print(a, b, c)
print('------------')
a[0] = -1
print(a is b)
print(a==b)
print(a is c)
print(a==c)
print(a, b, c)
```

```python
import copy
a = [[1,2], [3,4]] ## nested list
b = a.copy()
c = copy.deepcopy(a)
print(a)
print(b)
print(c)
print('------')
c[0][0]=200
print(a)
print(b)
print(c)
print('------')
b[0][0]=100
print(a)
print(b)
print(c)
print('------')
```
